```python
import zipfile
from pathlib import Path
from IPython.display import FileLink, display

files = [
    "/kaggle/working/symcode_results.csv",
    "/kaggle/working/symcode_predictions_full.json",
    "/kaggle/working/symcode_submission.json",
    "/kaggle/working/symcode_submission.jsonl",
    "/kaggle/working/symcode_partial_results.jsonl",
]

zip_path = "/kaggle/working/symcode_outputs.zip"

with zipfile.ZipFile(zip_path, "w") as z:
    for f in files:
        p = Path(f)
        if p.exists():
            z.write(p, arcname=p.name)
            print("added:", p.name)

display(FileLink(zip_path))
```
